# Despliegue del mapa final · Ruta 39 (Práctica 1, TIO)

Cuaderno de Colab nuevo para desplegar el mapa interactivo exportado desde Kepler.gl (`mapa_final_ruta39.html`). No hace falta instalar `keplergl` ni tener los CSV, porque los datos y la configuración van dentro del `.html`.

1. Ejecuta la celda y sube `mapa_final_ruta39.html`.
2. El mapa aparece debajo. Las capas se encienden y se apagan desde el panel *Layers* y los filtros se ajustan en *Filters*.

```python
import os, html
from IPython.display import HTML, display

FICHERO = 'mapa_final_ruta39.html'
if not os.path.exists(FICHERO):
    from google.colab import files
    subido = files.upload()
    FICHERO = list(subido.keys())[0]

with open(FICHERO, encoding='utf-8') as f:
    contenido = f.read()

display(HTML(f'<iframe srcdoc="{html.escape(contenido)}" width="100%" height="700" style="border:0"></iframe>'))
```
