## Carga .csv

```python
from keplergl import KeplerGl
import pandas as pd

from google.colab import output, files
output.enable_custom_widget_manager()

# Opción 1: Cargar archivo CSV desde Google Drive
uploaded = files.upload()

# Lee el archivo CSV subido (asegúrate de que el nombre del archivo sea correcto)
filename = list(uploaded.keys())[0]  # Toma el primer archivo subido

# 2. Leer el archivo CSV con los parámetros correctos para tu formato
# sep=';' -> Indica que el separador es punto y coma
# decimal=',' -> Indica que la coma es el separador decimal
df = pd.read_csv(filename, sep=';', decimal=',')
#df = pd.read_csv(filename, sep=';', decimal='.')

#df = pd.read_csv(filename)

# Opción 2: Si tienes el archivo en Google Drive, puedes montar el Drive
# from google.colab import drive
# drive.mount('/content/drive')
# df = pd.read_csv('/content/drive/MyDrive/nombre_del_archivo.csv')

# Opción 3: Cargar archivo CSV desde una URL
# url = 'https://dagoberto.es/datos.csv'
# df = pd.read_csv(url)

# Asegúrate de que las columnas de latitud y longitud estén correctamente nombradas
# Si no, puedes renombrarlas así:
# df = df.rename(columns={'nombre_lat': 'Latitude', 'nombre_lon': 'Longitude'})

# 2. Crear una instancia de KeplerGl
map_1 = KeplerGl(height=600)

# 3. Añadir el DataFrame al mapa
map_1.add_data(data=df, name='Datos_CSV')

# 4. Mostrar el mapa en Google Colab
from IPython.display import display
display(map_1)
```

### Entendiendo Coordenadas UTM

*   **UTM_X**: Corresponde a la coordenada Este (Easting) en UTM, similar a la "longitud" en un sistema cartesiano.
*   **UTM_Y**: Corresponde a la coordenada Norte (Northing) en UTM, similar a la "latitud" en un sistema cartesiano.

### Nomenclatura Común para Coordenadas Geográficas

#### Latitud (Latitude)

Nombres comunes:

*   `latitude`
*   `lat`
*   `y` (en algunos sistemas cartesianos)
*   `geo_lat`
*   `decimal_latitude`

#### Longitud (Longitude)

Nombres comunes:

*   `longitude`
*   `lon`
*   `lng`
*   `x` (en algunos sistemas cartesianos)
*   `geo_lon`
*   `decimal_longitude`
